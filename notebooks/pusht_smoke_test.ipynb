```python
import gymnasium as gym
import gym_pusht
import matplotlib.pyplot as plt
import json
import numpy as np


env = gym.make("gym_pusht/PushT-v0", render_mode = "rgb_array")
obs, info = env.reset(seed = 0)
obs, reward, terminated, truncated, info = env.step(env.action_space.sample())
frame = env.render()
env.close()

plt.figure()
plt.imshow(frame)
plt.axis("off")
plt.show()
print("frame shape:", getattr(frame, "shape", None))

serializable_info = {k: v.tolist() if isinstance(v, np.ndarray) else v for k, v in info.items()}
print(obs, reward, terminated, truncated, json.dumps(serializable_info, indent = 4), sep = "\n\n")
```
